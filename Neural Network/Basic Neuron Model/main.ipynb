```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# AND input
X = [[0,0],[0,1],[1,0],[1,1]]
```

```python
# AND output
Y = [[0],[0],[0],[1]]
```

```python
for i in range(len(X)):
    print(X[i],Y[i])
```

```python
pred = []
for i in range(len(X)):
    sigma = (3 * X[i][0]) + (2 * X[i][1])
    pred.append(1 if sigma >= 4 else 0)
pred
```

```python
# plot

# coordination of the data
x_coords = [data_point[0] for data_point in X]
y_coords = [data_point[1] for data_point in X]

# coordination of the line
x_values = np.linspace(0, 2, 100)
y_values = (4- 3 * x_values) / 2

plt.scatter(x_coords, y_coords, c=pred)
plt.plot(x_values, y_values, color='red')
plt.plot()
```
