```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Conditional Combination
X = [[0,0],[0,1],[1,0],[1,1]]
Y = [[1],[1],[0],[1]]
```

```python
for i in range(len(X)):
  print(X[i],Y[i])
```

```python
pred = []
for i in range(len(X)):
  sigma = (2 * X[i][0]) + (-2 * X[i][1])
  pred.append(1 if sigma < 1 else 0)
pred
```

```python
# Plot

x1_coords = [point[0] for point in X]
x2_coords = [point[1] for point in X]

x_values = np.linspace(0,1,100)
y_values = ((2 * x_values) - 1) / (2)

plt.scatter(x1_coords, x2_coords, c=pred, cmap='viridis')
plt.plot(x_values, y_values, color='red')
plt.title('Conditional Combination')
plt.xlabel('X1')
plt.ylabel('X2')
```
